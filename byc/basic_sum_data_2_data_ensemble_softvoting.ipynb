```python
# 모든 열을 다 보여주도록 설정
pd.set_option('display.max_columns', None)
```

```python
# Q1: 클래스 확률 비교 후 더 큰 쪽 클래스 선택
submission_df['Q1_pred'] = submission_df[['Q1_class0_proba_pv', 'Q1_class1_proba_pv']].idxmax(axis=1)

# 컬럼명에서 'classX_proba_pv' 중 'X'만 추출해서 정수형 클래스값으로 변환
submission_df['Q1_pred'] = submission_df['Q1_pred'].str.extract(r'class(\d+)_proba_pv').astype(int)
```
